## 5. Data Analysis

### 5.1 Average Final Grade

Calculate the average final grade of all students using the G3 column.

```python
average_grade = df['G3'].mean()

print("Average Final Grade:", average_grade)
```

**Observation:**

The average final grade of students is approximately 10.42 out of 20.

### 5.2 Students Scoring Above 15

Calculate the number of students who obtained a final grade greater than 15.

```python
students_above_15 = (df['G3'] > 15).sum()

print("Students scoring above 15:", students_above_15)
```

**Observation:**

A total of 40 students scored above 15 in their final mathematics examination.

### 5.3 Relationship Between Study Time and Final Grades

We calculate the correlation between students' study time and their final grades (G3) to understand whether study time is associated with academic performance.

```python
correlation = df['studytime'].corr(df['G3'])

print("Study Time and Final Grade Correlation:", correlation)
```

**Observation:**

The correlation between study time and final grades is approximately 0.098, indicating a very weak positive relationship. This suggests that study time alone has a weak linear association with students' academic performance.

### 5.4 Gender-wise Performance Analysis

We compare the average final grades of male and female students to identify which group performs better on average.

```python
gender_performance = df.groupby('sex')['G3'].mean()

print(gender_performance)
```

```python
print("Gender with higher average performance:",
      gender_performance.idxmax())
```

**Observation:**

The average final grade of female students is approximately 9.97, while male students have an average of 10.91. Therefore, male students perform slightly better on average in this dataset.

## 6. Data Visualization

### 6.1 Histogram of Final Grades

A histogram is used to visualize the distribution of students' final grades (G3).

```python
plt.figure(figsize=(8, 5))

plt.hist(df['G3'], bins=10, edgecolor='black')

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

**Observation:**

The histogram shows the distribution of final grades among 395 students. Most students scored between 10 and 12. A considerable number of students scored between 8 and 15, while fewer students achieved grades above 18. Some students also received a final grade of 0.

### 6.2 Scatterplot: Study Time vs Final Grades

A scatterplot is used to visualize the relationship between students' study time and their final grades.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['studytime'], df['G3'], alpha=0.6)

plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

**Observation:**

The scatterplot shows the relationship between study time and final grades. Students with different study-time categories have varying grades. There is no strong visible relationship between study time and academic performance, which is consistent with the very weak positive correlation of approximately 0.098.

### 6.3 Bar Chart: Male vs Female Average Score

A bar chart is used to compare the average final grades of male and female students.

```python
plt.figure(figsize=(7, 5))

sns.barplot(x='sex', y='G3', data=df)

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.xticks([0, 1], ['Female', 'Male'])

plt.show()
```

**Observation:**

The bar chart compares the average final grades of male and female students. Female students scored an average of 9.97, while male students scored an average of 10.91. Therefore, male students performed slightly better on average in this dataset.
